#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate(root: TreeNode) -> int:
    def dfs(node, seen):
        if not node:
            return False
        if node.val in seen:
            return True
        seen.add(node.val)
        return dfs(node.left, seen) or dfs(node.right, seen)

    return int(dfs(root, set()))
```

```python
# Client Code
# Build a sample binary tree:
#         5
#        / \
#       3   7
#      / \   \
#     2   3   9   ← duplicate value: 3
root = TreeNode(5)
root.left = TreeNode(3)
root.right = TreeNode(7)
root.left.left = TreeNode(2)
root.left.right = TreeNode(3)  # duplicate
root.right.right = TreeNode(9)

# Call the function
result = is_duplicate(root)

# Output the result
if result:
    print("Duplicate found in the tree.")
else:
    print("No duplicates in the tree.")
```

```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    insert_pos = 0 

  
    for num in nums:
        if num != 0:
            nums[insert_pos] = num
            insert_pos += 1

    
    while insert_pos < len(nums):
        nums[insert_pos] = 0
        insert_pos += 1

    return nums

nums = [0, 1, 0, 3, 12]
print(f'Input: {nums}')
print(f'Output: {move_zeros_to_end(nums=nums)}')

nums = [4, 0, 5, 0, 0, 6]
print(f'Input: {nums}')
print(f'Output: {move_zeros_to_end(nums=nums)}')

nums = [0, 1, 0, 4, 0, 16]
print(f'Input: {nums}')
print(f'Output: {move_zeros_to_end(nums=nums)}')
```


-   Explain the problem’s time and space complexity in your own words.


```python
'''
Time Complexity: O(n)

-   The function processes each element of the list exactly once.
-   It has two loops, but they are sequential, not nested:
    1.  First loop: Iterates through all elements to move non-zero values forward.
    2.  Second loop: Fills in zeros from the insert_pos to the end.

-   Even in the worst case (e.g., all elements are non-zero or all are zero), 
    the total number of operations is at most 2n.
-   In Big-O notation, constants are dropped → O(2n) = O(n)

Space Complexity: O(1)

-   The function modifies the list in-place.
-   It uses only a few scalar variables (insert_pos, num) — no additional data structures.
-   No matter how large the input list is, the extra memory used remains constant.
-   Therefore, space complexity is O(1)

Summary:

Metric   Value   Explaination
Time     O(n)    Each element is visited once.
Space    O(1)    No extra space used.
'''
```

```python
'''
Strengths:

1.  Correctness:
    -   The function correctly moves all zeros to the end while preserving the order 
        of non-zero elements.
    -   It handles edge cases like empty lists, all zeros, or no zeros.

2.  In-place operation:
    -   It modifies the input list directly, which is memory-efficient.

3.  Time and space efficiency:
    -   Time: O(n) — each element is visited once.
    -   Space: O(1) — no extra data structures are used.

4.  Simplicity:
    -   The logic is clean and easy to follow.
 
Areas for Improvement:

1.  Unnecessary writes:
    -   If the list has many non-zero elements already in place, the function 
        still overwrites them unnecessarily.
    -   This can be optimized by checking if insert_pos != current_index before writing.
    -   Improved version:

        for i in range(len(nums)):
            if nums[i] != 0:
                if i != insert_pos:
                    nums[insert_pos] = nums[i]
                insert_pos += 1

2.  Function Naming:
    -   The name move_zeros_to_end is descriptive, but could be clarified that it is 
        an in-place operation in the docstring.

3.  Type hinting and documentation:
    -   A brief docstring would improve clarity for future readers or collaborators.

Suggested Final Version:

from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    """
    Moves all zeros in the list to the end in-place,
    preserving the order of non-zero elements.
    """
    insert_pos = 0

    for i in range(len(nums)):
        if nums[i] != 0:
            if i != insert_pos:
                nums[insert_pos] = nums[i]
            insert_pos += 1

    while insert_pos < len(nums):
        nums[insert_pos] = 0
        insert_pos += 1

    return nums
'''
```
